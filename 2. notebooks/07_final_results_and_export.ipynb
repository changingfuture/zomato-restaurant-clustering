# Final Results & Export

In this notebook, I will consolidate the final results of the restaurant clustering project, summarize the characteristics of the identified clusters, document the key findings, and verify the final project outputs.

## 1. Imports, Setup & Data Loading

I will import the necessary libraries, establish the project paths, and load the final clustered restaurant dataset.

This dataset contains the restaurant information, engineered features, and cluster assignments generated by the final K-Means model.

```python
import pandas as pd 
import numpy as np 
from pathlib import Path

# Project Setup
project_root = Path.cwd().parent
processed_data_path = project_root / "1. data" / "processed"
models_path = project_root / "4. models"

# Loading Final Clustered Dataset
final_results = pd.read_csv(
    processed_data_path / "final_restaurant_clusters.csv"
)
display(final_results.head())
print("Dataset Shape: ", final_results.shape)
print("\nCluster Distribution:")
display(final_results["Clusters"].value_counts().sort_index())
```

## 2. Final Clustering Model

K-Means clustering was selected as the final clustering algorithm for this project.

Multiple cluster configurations were evaluated using the Elbow Method, Silhouette Score, Davies-Bouldin Index, and Calinski-Harabasz Score.

After comparing K = 2, K = 3, and K = 4, the K = 2 configuration was selected as the final model because it achieved the highest Silhouette Score, the lowest Davies-Bouldin Index, and the highest Calinski-Harabasz Score among the evaluated configurations.

The selected model was fitted using 2 clusters, with random_state = 42 and n_init = 10.

```python
# Final Model Configuration
final_model_parameters = {
    "Algorithm" : "K-Means",
    "Number of Clusters (K)" : 2,
    "Random State" : 42,
    "Number of Initializations (n_init)" : 10
}
display(pd.DataFrame([final_model_parameters]))
```

```python
# Final Model Evaluation 
final_model_metrics = pd.DataFrame({
    "Metric": [
        "Silhouette Score",
        "Davies-Bouldin Index",
        "Calinski-Harabasz",
        "Inertia"
    ], 
    "Score" : [
        0.207281, 1.701194, 34.01734, 461.720760
    ]
})
display(final_model_metrics.set_index("Metric"))
```

## 3. Final Cluster Profiles

The two clusters identified by the K-Means model exhibit differences in restaurant pricing, customer ratings, review characteristics, and customer engagement.

I will consolidate the key numerical characteristics of both clusters into a comparative profile to summarize the findings of the clustering analysis.

```python
# Final Cluster Profiles
cluster_profiles = pd.DataFrame({
    "Cluster 0" : [
        55, 
        533.64,
        3.33,
        1.53,
        23.7,
        0.32],
    
    "Cluster 1" : [
        50,
        1222.00,
        3.88,
        1.22,
        44.1,
        1.22]
}, index= [
    "Number of Restaurants",
    "Average Cost",
    "Average Rating",
    "Rating Variability",
    "Median Review Length (Words)",
    "Average Pictures per Review"
])
display(cluster_profiles)
```

### 3.1 Cluster Interpretation

**Cluster 0** represents relatively affordable restaurants with lower average ratings, greater rating variability, shorter customer reviews, and fewer pictures per review. This cluster reflects restaurants where customer ratings and engagement are comparatively lower.

**Cluster 1** represents relatively higher-cost restaurants with higher average ratings, more consistent customer ratings, longer customer reviews, and greater picture engagement. This cluster reflects restaurants with comparatively stronger customer engagement and more positive rating patterns.

Although these clusters exhibit noticeable differences, they should be interpreted as distinct restaurant profiles rather than strict quality categories.

## 4. Key Findings & Final Conclusions

The clustering analysis identified two distinct restaurant groups based on pricing, cuisine characteristics, customer ratings, review patterns, and picture engagement.

- **Cluster 0 – Affordable Restaurants:** This cluster consists of relatively lower-cost restaurants with lower average ratings, greater rating variability, shorter reviews, and lower picture engagement.
- **Cluster 1 – Higher-Cost Restaurants:** This cluster consists of relatively higher-cost restaurants with higher average ratings, more consistent ratings, longer reviews, and greater picture engagement.

The analysis suggests that restaurant cost is associated with differences in customer ratings and engagement patterns within the dataset. However, these clusters are not strictly defined by price or restaurant quality, as cuisine offerings and restaurant characteristics overlap across the two groups.

The final K-Means model with two clusters was selected based on the comparative evaluation of clustering metrics. Although it performed better than the other tested configurations, the relatively low Silhouette Score indicates that the clusters are not strongly separated. Therefore, the results should be interpreted as broad restaurant profiles rather than rigid categories.

Overall, the project demonstrates how clustering can be used to identify patterns in restaurant data and provide insights into restaurant pricing, customer ratings, review characteristics, and customer engagement.
